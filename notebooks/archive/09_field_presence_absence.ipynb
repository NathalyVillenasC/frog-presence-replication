# 09 — Field presence/absence inference

## Objective

This notebook is the final large methodological stage of the paper replication:

**Dataset-2 pond WAV → frozen segmentation → original unfiltered candidate → 20 PLP-RASTA → frozen
GMMs → final LLR verification → presence/absence vector**

Everything developed in notebooks 06B–08D is now treated as frozen.

### Frozen detector

- feature: **20 PLP-RASTA**
- one diagonal GMM per deployable species
- `M = 4`
- winning hypothesis = largest mean frame log-likelihood
- verification:

\[
LLR_k(X)
=
s_k(X)
-
\operatorname{median}_{j\neq k}s_j(X)
\]

- accept the winning hypothesis only when its final threshold is passed.

### Frozen segmentation

- consecutive **30 s** blocks;
- Table-1 species frequency bands;
- FIR used only for endpoint localization;
- **10 ms non-overlapping STE**;
- moving average **MA = 12**;
- **3 above / 3 below** endpoint persistence;
- species-specific `C` learned on Dataset 1;
- `f04` only: **20 ms symmetric endpoint padding**;
- every classified waveform is taken from the **original unfiltered recording**.

`[PAPER]` Presence in one field sample is declared when at least one call of the species is accepted.

---

## Important replication limitations

1. The paper evaluates 10 species. Our recovered `f09` labelled corpus is insufficient, so the deployed
   detector contains **9 species**. `f09` is reported as `NA`, never as biological absence.
2. The paper does not publish the segmentation constant `C` or FIR order. Our final values are documented
   reconstruction/development decisions.
3. The paper ultimately used m-FCC/M=64; our recording-separated experiments selected
   **PLP-RASTA/M=4**. This is a documented project divergence, not a claim about the paper.
4. The paper manually labelled 18 field samples with complete 10-species vectors. Unless those exact
   complete labels are recoverable locally, this notebook **does not invent test metrics** from partial
   notes.
5. Thresholds and segmentation parameters are not retuned on ponds. The field corpus is finally allowed
   to be what a test corpus is supposed to be, a concept humanity occasionally remembers.

## 5. Inventory Dataset 2

`[PAPER]` The test material is described as 141 nominally 10-minute samples digitized at 48 kHz.

The recovered collection is audited rather than forced into that description. A WAV is eligible when:

- it can be read by libsndfile;
- sample rate is 48 kHz;
- subtype is PCM.

Duration is reported as observed. A non-600-second file is not silently discarded.

```python
def infer_pond(path):
    text = " ".join(
        part.lower()
        for part in path.parts
    )

    p1_patterns = (
        r"\bp1\b",
        r"pond\s*1",
        r"poza\s*1",
    )

    p2_patterns = (
        r"\bp2\b",
        r"pond\s*2",
        r"poza\s*2",
    )

    if any(
        re.search(
            pattern,
            text,
        )
        for pattern
        in p1_patterns
    ):
        return "P1"

    if any(
        re.search(
            pattern,
            text,
        )
        for pattern
        in p2_patterns
    ):
        return "P2"

    return "unknown"


def infer_month(path):
    text = " ".join(
        part.lower()
        for part in path.parts
    )

    month_patterns = {
        "February": (
            r"feb",
            r"febrero",
            r"02[_\- ]",
        ),
        "April": (
            r"apr",
            r"abril",
            r"04[_\- ]",
        ),
        "July": (
            r"jul",
            r"julio",
            r"07[_\- ]",
        ),
        "August": (
            r"aug",
            r"agosto",
            r"08[_\- ]",
        ),
        "September": (
            r"sep",
            r"sept",
            r"septiembre",
            r"09[_\- ]",
        ),
    }

    for month, patterns in month_patterns.items():
        if any(
            re.search(
                pattern,
                text,
            )
            for pattern
            in patterns
        ):
            return month

    return "unknown"


def sample_id_from_path(path):
    relative = path.relative_to(
        PONDS_CALL_DIR
    ).with_suffix("")

    return "__".join(
        relative.parts
    )


inventory_rows = []

wav_paths = sorted(
    path
    for path
    in PONDS_CALL_DIR.rglob("*")
    if path.suffix.lower() == ".wav"
)

for path in wav_paths:
    row = {
        "sample_id":
            sample_id_from_path(
                path
            ),
        "path":
            str(path),
        "relative_path":
            str(
                path.relative_to(
                    PONDS_CALL_DIR
                )
            ),
        "pond":
            infer_pond(
                path
            ),
        "month":
            infer_month(
                path
            ),
        "readable":
            False,
        "sample_rate":
            np.nan,
        "frames":
            np.nan,
        "duration_seconds":
            np.nan,
        "format":
            None,
        "subtype":
            None,
        "eligible":
            False,
        "reason":
            None,
    }

    try:
        info = sf.info(
            path
        )

        row.update({
            "readable":
                True,
            "sample_rate":
                int(
                    info.samplerate
                ),
            "frames":
                int(
                    info.frames
                ),
            "duration_seconds":
                float(
                    info.duration
                ),
            "format":
                str(
                    info.format
                ),
            "subtype":
                str(
                    info.subtype
                ),
        })

        reasons = []

        if int(
            info.samplerate
        ) != 48_000:
            reasons.append(
                f"sample_rate={info.samplerate}"
            )

        if not str(
            info.subtype
        ).startswith(
            "PCM_"
        ):
            reasons.append(
                f"subtype={info.subtype}"
            )

        row[
            "eligible"
        ] = (
            len(reasons)
            == 0
        )

        row[
            "reason"
        ] = (
            "; ".join(
                reasons
            )
            if reasons
            else "ok"
        )

    except Exception as exc:
        row[
            "reason"
        ] = repr(
            exc
        )

    inventory_rows.append(
        row
    )

field_inventory = pd.DataFrame(
    inventory_rows
)

if (
    not field_inventory.empty
    and not field_inventory[
        "sample_id"
    ].is_unique
):
    raise RuntimeError(
        "Dataset-2 sample identifiers are not unique."
    )

field_inventory.to_csv(
    RESULTS_DIR
    / "field_inventory.csv",
    index=False,
)

print("WAV files found :", len(field_inventory))
print(
    "Eligible WAVs   :",
    int(
        field_inventory[
            "eligible"
        ].sum()
    )
    if len(field_inventory)
    else 0,
)

if len(
    field_inventory
):
    display(
        field_inventory[
            [
                "sample_id",
                "pond",
                "month",
                "sample_rate",
                "duration_seconds",
                "subtype",
                "eligible",
                "reason",
            ]
        ]
        .head(
            30
        )
        .round(3)
    )

eligible_inventory = (
    field_inventory[
        field_inventory[
            "eligible"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)

if len(
    eligible_inventory
) != 141:
    warnings.warn(
        f"Paper reports 141 nominal field samples; "
        f"recovered eligible inventory contains {len(eligible_inventory)}."
    )

if len(
    eligible_inventory
):
    print(
        "Eligible duration:",
        round(
            eligible_inventory[
                "duration_seconds"
            ].sum()
            / 3600.0,
            3,
        ),
        "hours",
    )
```

## 6. Frozen segmentation functions

```python
BLOCK_SECONDS = float(
    segmentation_config[
        "paper_frontend"
    ][
        "block_seconds"
    ]
)

STE_FRAME_SECONDS = float(
    segmentation_config[
        "paper_frontend"
    ][
        "ste_frame_seconds"
    ]
)

STE_MA_LENGTH = int(
    segmentation_config[
        "paper_frontend"
    ][
        "moving_average_length"
    ]
)

ENDPOINT_RUN = int(
    segmentation_config[
        "paper_frontend"
    ][
        "endpoint_run_length"
    ]
)

FIR_NUMTAPS = int(
    segmentation_config[
        "replication_decisions"
    ][
        "fir_numtaps"
    ]
)


def to_mono(audio):
    audio = np.asarray(
        audio,
        dtype=np.float64,
    )

    if audio.ndim == 1:
        return audio

    if audio.ndim == 2:
        return audio.mean(
            axis=1
        )

    raise ValueError(
        f"Unsupported audio shape: {audio.shape}"
    )


_FIR_CACHE = {}


def bandpass_for_segmentation(
    audio,
    sr,
    low_hz,
    high_hz,
):
    key = (
        int(sr),
        float(low_hz),
        float(high_hz),
        FIR_NUMTAPS,
    )

    if key not in _FIR_CACHE:
        _FIR_CACHE[
            key
        ] = firwin(
            numtaps=FIR_NUMTAPS,
            cutoff=[
                low_hz,
                high_hz,
            ],
            pass_zero=False,
            fs=sr,
        )

    return fftconvolve(
        audio,
        _FIR_CACHE[
            key
        ],
        mode="same",
    )


def short_time_energy(
    audio,
    sr,
):
    frame_samples = int(
        round(
            sr
            * STE_FRAME_SECONDS
        )
    )

    n_frames = (
        len(audio)
        // frame_samples
    )

    if n_frames == 0:
        return (
            np.empty(
                0,
                dtype=float,
            ),
            frame_samples,
        )

    trimmed = audio[
        :
        n_frames
        * frame_samples
    ]

    frames = trimmed.reshape(
        n_frames,
        frame_samples,
    )

    energy = np.sum(
        frames
        * frames,
        axis=1,
    )

    return (
        energy,
        frame_samples,
    )


def moving_average_same(
    values,
):
    values = np.asarray(
        values,
        dtype=float,
    )

    if len(values) == 0:
        return values.copy()

    kernel = (
        np.ones(
            STE_MA_LENGTH,
            dtype=float,
        )
        / STE_MA_LENGTH
    )

    return np.convolve(
        values,
        kernel,
        mode="same",
    )


def energy_to_db(
    energy,
):
    return (
        10.0
        * np.log10(
            np.maximum(
                np.asarray(
                    energy,
                    dtype=float,
                ),
                np.finfo(
                    float
                ).tiny,
            )
        )
    )


def endpoint_threshold_db(
    ste_db,
    C,
):
    ste_db = np.asarray(
        ste_db,
        dtype=float,
    )

    if len(ste_db) == 0:
        return np.nan

    mean_db = float(
        ste_db.mean()
    )

    max_db = float(
        ste_db.max()
    )

    return (
        mean_db
        + (
            max_db
            - mean_db
        )
        / C
    )


def find_endpoint_runs(
    ste_db,
    threshold_db,
):
    ste_db = np.asarray(
        ste_db,
        dtype=float,
    )

    segments = []

    in_segment = False
    start_idx = None

    i = 0

    while (
        i
        <= len(ste_db)
        - ENDPOINT_RUN
    ):
        current = ste_db[
            i:
            i
            + ENDPOINT_RUN
        ]

        if not in_segment:
            if np.all(
                current
                > threshold_db
            ):
                start_idx = i
                in_segment = True
                i += ENDPOINT_RUN
                continue

        else:
            if np.all(
                current
                < threshold_db
            ):
                end_idx = i

                if (
                    start_idx
                    is not None
                    and end_idx
                    > start_idx
                ):
                    segments.append(
                        (
                            start_idx,
                            end_idx,
                        )
                    )

                start_idx = None
                in_segment = False
                i += ENDPOINT_RUN
                continue

        i += 1

    if (
        in_segment
        and start_idx
        is not None
    ):
        segments.append(
            (
                start_idx,
                len(
                    ste_db
                ),
            )
        )

    return segments


def merge_intervals(
    intervals,
):
    intervals = sorted(
        (
            int(start),
            int(end),
        )
        for start, end
        in intervals
        if end > start
    )

    if not intervals:
        return []

    merged = [
        list(
            intervals[0]
        )
    ]

    for start, end in intervals[
        1:
    ]:
        previous = merged[
            -1
        ]

        if start <= previous[1]:
            previous[1] = max(
                previous[1],
                end,
            )
        else:
            merged.append(
                [
                    start,
                    end,
                ]
            )

    return [
        (
            start,
            end,
        )
        for start, end
        in merged
    ]
```

## 7. Exact 20-dimensional PLP-RASTA frontend used by the frozen models

```python
PRE_EMPHASIS = 0.99
FEATURE_FRAME_SECONDS = 0.020
FEATURE_OVERLAP = 0.75
NFFT = 1024

N_BARK_FILTERS = 24
PLP_ORDER = 20
N_PLP_CEPSTRA = 20


def pre_emphasise(audio):
    audio = np.asarray(
        audio,
        dtype=np.float64,
    )

    if len(audio) == 0:
        return audio.copy()

    out = np.empty_like(
        audio
    )

    out[0] = audio[0]

    out[1:] = (
        audio[1:]
        - PRE_EMPHASIS
        * audio[:-1]
    )

    return out


def frame_audio(
    audio,
    sr,
):
    frame_length = int(
        round(
            FEATURE_FRAME_SECONDS
            * sr
        )
    )

    hop_length = int(
        round(
            FEATURE_FRAME_SECONDS
            * (
                1.0
                - FEATURE_OVERLAP
            )
            * sr
        )
    )

    if len(audio) < frame_length:
        return np.empty(
            (
                0,
                frame_length,
            ),
            dtype=float,
        )

    n_frames = (
        1
        + (
            len(audio)
            - frame_length
        )
        // hop_length
    )

    starts = (
        np.arange(
            n_frames
        )
        * hop_length
    )

    return np.stack([
        audio[
            start:
            start
            + frame_length
        ]
        for start
        in starts
    ])


def frame_power_spectrum(
    audio,
    sr,
):
    emphasized = pre_emphasise(
        audio
    )

    frames = frame_audio(
        emphasized,
        sr,
    )

    if len(frames) == 0:
        return (
            np.empty(
                (
                    0,
                    NFFT // 2 + 1,
                )
            ),
            frames,
        )

    window = np.hamming(
        frames.shape[1]
    )

    spectrum = np.fft.rfft(
        frames
        * window[
            None,
            :
        ],
        n=NFFT,
        axis=1,
    )

    power = (
        np.abs(
            spectrum
        ) ** 2
    ) / NFFT

    return (
        power,
        frames,
    )


def hz_to_bark(hz):
    hz = np.asarray(
        hz,
        dtype=float,
    )

    return (
        13.0
        * np.arctan(
            0.00076
            * hz
        )
        + 3.5
        * np.arctan(
            (
                hz
                / 7500.0
            ) ** 2
        )
    )


def bark_to_hz(bark):
    bark = np.asarray(
        bark,
        dtype=float,
    )

    hz_grid = np.linspace(
        0.0,
        24_000.0,
        200_001,
    )

    bark_grid = hz_to_bark(
        hz_grid
    )

    return np.interp(
        bark,
        bark_grid,
        hz_grid,
    )


_BARK_BANK_CACHE = {}


def bark_filterbank(
    sr,
    n_fft=NFFT,
    n_filters=N_BARK_FILTERS,
):
    key = (
        int(sr),
        int(n_fft),
        int(n_filters),
    )

    if key in _BARK_BANK_CACHE:
        return _BARK_BANK_CACHE[
            key
        ]

    max_bark = float(
        hz_to_bark(
            sr / 2.0
        )
    )

    bark_points = np.linspace(
        0.0,
        max_bark,
        n_filters
        + 2,
    )

    hz_points = bark_to_hz(
        bark_points
    )

    fft_freqs = np.fft.rfftfreq(
        n_fft,
        d=1.0 / sr,
    )

    bank = np.zeros(
        (
            n_filters,
            len(
                fft_freqs
            ),
        ),
        dtype=float,
    )

    centers = []

    for i in range(
        n_filters
    ):
        left = hz_points[
            i
        ]

        center = hz_points[
            i + 1
        ]

        right = hz_points[
            i + 2
        ]

        centers.append(
            center
        )

        left_mask = (
            (
                fft_freqs
                >= left
            )
            & (
                fft_freqs
                <= center
            )
        )

        right_mask = (
            (
                fft_freqs
                >= center
            )
            & (
                fft_freqs
                <= right
            )
        )

        if center > left:
            bank[
                i,
                left_mask,
            ] = (
                fft_freqs[
                    left_mask
                ]
                - left
            ) / (
                center
                - left
            )

        if right > center:
            bank[
                i,
                right_mask,
            ] = (
                right
                - fft_freqs[
                    right_mask
                ]
            ) / (
                right
                - center
            )

    result = (
        bank,
        np.asarray(
            centers
        ),
    )

    _BARK_BANK_CACHE[
        key
    ] = result

    return result


def rasta_filter_log_spectrum(
    log_spectrum,
):
    numerator = np.array([
        0.2,
        0.1,
        0.0,
        -0.1,
        -0.2,
    ])

    denominator = np.array([
        1.0,
        -0.94,
    ])

    return lfilter(
        numerator,
        denominator,
        log_spectrum,
        axis=0,
    )


def equal_loudness_curve(
    freq_hz,
):
    f2 = np.asarray(
        freq_hz,
        dtype=float,
    ) ** 2

    numerator = (
        (
            f2
            + 1.6e5
        )
        * (
            f2 ** 2
        )
    )

    denominator = (
        (
            f2
            + 1.44e6
        )
        * (
            (
                f2
                + 9.61e6
            ) ** 2
        )
    )

    curve = (
        numerator
        / np.maximum(
            denominator,
            np.finfo(
                float
            ).eps,
        )
    )

    return np.maximum(
        curve,
        np.finfo(
            float
        ).eps,
    )


def auditory_spectrum_to_lpc(
    spectrum,
    order=PLP_ORDER,
):
    spectrum = np.asarray(
        spectrum,
        dtype=float,
    )

    n_ifft = 2 * (
        len(
            spectrum
        )
        - 1
    )

    autocorr = np.fft.irfft(
        spectrum,
        n=n_ifft,
    )

    autocorr = np.real(
        autocorr[
            :
            order
            + 1
        ]
    )

    autocorr[0] = max(
        autocorr[0],
        1e-10,
    )

    r0 = autocorr[0]

    autocorr[0] += (
        1e-6
        * abs(
            r0
        )
    )

    try:
        coeff = solve_toeplitz(
            (
                autocorr[
                    :
                    order
                ],
                autocorr[
                    :
                    order
                ],
            ),
            -autocorr[
                1:
                order
                + 1
            ],
        )

    except Exception:
        coeff = np.zeros(
            order,
            dtype=float,
        )

    lpc = np.concatenate([
        [
            1.0
        ],
        coeff,
    ])

    error = float(
        autocorr[0]
        + np.dot(
            autocorr[
                1:
                order
                + 1
            ],
            coeff,
        )
    )

    return (
        lpc,
        max(
            error,
            1e-12,
        ),
    )


def lpc_to_cepstra(
    lpc,
    error,
    n_cepstra=N_PLP_CEPSTRA,
):
    order = len(
        lpc
    ) - 1

    c = np.zeros(
        n_cepstra,
        dtype=float,
    )

    c[0] = np.log(
        max(
            error,
            1e-12,
        )
    )

    for n in range(
        1,
        n_cepstra,
    ):
        total = 0.0

        k_start = max(
            1,
            n
            - order,
        )

        for k in range(
            k_start,
            n,
        ):
            a_index = (
                n
                - k
            )

            if (
                a_index
                <= order
                and k
                < len(c)
            ):
                total += (
                    (
                        k
                        / n
                    )
                    * c[k]
                    * lpc[
                        a_index
                    ]
                )

        if n <= order:
            c[n] = (
                -lpc[n]
                - total
            )
        else:
            c[n] = (
                -total
            )

    return c


def plp_rasta_20(
    audio,
    sr,
):
    power, frames = frame_power_spectrum(
        audio,
        sr,
    )

    if len(frames) == 0:
        return np.empty(
            (
                0,
                N_PLP_CEPSTRA,
            )
        )

    bark_bank, centers_hz = bark_filterbank(
        sr
    )

    critical = (
        power
        @ bark_bank.T
    )

    critical = np.maximum(
        critical,
        np.finfo(
            float
        ).eps,
    )

    log_critical = np.log(
        critical
    )

    rasta_log = (
        rasta_filter_log_spectrum(
            log_critical
        )
    )

    rasta_energy = np.exp(
        rasta_log
    )

    eql = equal_loudness_curve(
        centers_hz
    )

    post_auditory = (
        rasta_energy
        * eql[
            None,
            :
        ]
    )

    compressed = np.cbrt(
        np.maximum(
            post_auditory,
            np.finfo(
                float
            ).eps,
        )
    )

    features_out = []

    for auditory_frame in compressed:
        lpc, error = (
            auditory_spectrum_to_lpc(
                auditory_frame,
                order=PLP_ORDER,
            )
        )

        features_out.append(
            lpc_to_cepstra(
                lpc,
                error,
                n_cepstra=N_PLP_CEPSTRA,
            )
        )

    return np.asarray(
        features_out,
        dtype=float,
    )
```

## 8. Frozen detector scoring

```python
def score_candidate_features(
    X,
):
    X = np.asarray(
        X,
        dtype=float,
    )

    if (
        X.ndim != 2
        or X.shape[1] != 20
        or len(X) == 0
        or not np.isfinite(
            X
        ).all()
    ):
        return None

    scores = {
        code_name:
            float(
                model.score(
                    X
                )
            )
        for code_name, model
        in MODELS.items()
    }

    hypothesis = max(
        scores,
        key=scores.get,
    )

    alternative_scores = [
        score
        for code_name, score
        in scores.items()
        if code_name
        != hypothesis
    ]

    llr = (
        scores[
            hypothesis
        ]
        - float(
            np.median(
                alternative_scores
            )
        )
    )

    threshold = float(
        THRESHOLDS[
            hypothesis
        ]
    )

    accepted = (
        llr
        >= threshold
    )

    return {
        "scores":
            scores,
        "hypothesis":
            hypothesis,
        "hypothesis_score":
            scores[
                hypothesis
            ],
        "llr":
            llr,
        "threshold":
            threshold,
        "accepted":
            bool(
                accepted
            ),
        "prediction":
            (
                hypothesis
                if accepted
                else None
            ),
    }
```

## 9. Segment one field file

The file is read in 30-second blocks for segmentation. Candidate timestamps are stored globally.

After all blocks have been segmented:

1. endpoint padding is applied;
2. intervals are merged **within their segmentation band**;
3. candidate audio is reread directly from the original WAV;
4. PLP-RASTA and the frozen detector are applied.

This avoids holding an entire 10-minute recording in RAM while still allowing the 20 ms `f04` padding
to cross a 30-second boundary.

```python
def segment_field_file(
    path,
):
    path = Path(
        path
    )

    info = sf.info(
        path
    )

    sr = int(
        info.samplerate
    )

    total_samples = int(
        info.frames
    )

    block_samples = int(
        round(
            BLOCK_SECONDS
            * sr
        )
    )

    raw_candidate_rows = []

    with sf.SoundFile(
        path,
        "r",
    ) as audio_file:

        n_blocks = int(
            np.ceil(
                total_samples
                / block_samples
            )
        )

        for block_index in range(
            n_blocks
        ):
            block_start = (
                block_index
                * block_samples
            )

            audio_file.seek(
                block_start
            )

            original_block = audio_file.read(
                frames=block_samples,
                dtype="float64",
                always_2d=False,
            )

            original_block = to_mono(
                original_block
            )

            if len(
                original_block
            ) == 0:
                continue

            if (
                BLOCK_PROGRESS_EVERY
                and (
                    block_index
                    % BLOCK_PROGRESS_EVERY
                    == 0
                )
            ):
                print(
                    f"    block {block_index + 1}/{n_blocks}"
                )

            for segmentation_code in CLASSES:
                low_hz, high_hz = (
                    BANDS_HZ[
                        segmentation_code
                    ]
                )

                filtered = (
                    bandpass_for_segmentation(
                        original_block,
                        sr,
                        low_hz,
                        high_hz,
                    )
                )

                ste, frame_samples = (
                    short_time_energy(
                        filtered,
                        sr,
                    )
                )

                ste_db = energy_to_db(
                    moving_average_same(
                        ste
                    )
                )

                if len(
                    ste_db
                ) == 0:
                    continue

                C = C_BY_CODE[
                    segmentation_code
                ]

                threshold_db = (
                    endpoint_threshold_db(
                        ste_db,
                        C,
                    )
                )

                frame_segments = (
                    find_endpoint_runs(
                        ste_db,
                        threshold_db,
                    )
                )

                for (
                    start_frame,
                    end_frame,
                ) in frame_segments:
                    start_sample = (
                        block_start
                        + start_frame
                        * frame_samples
                    )

                    end_sample = min(
                        total_samples,
                        block_start
                        + end_frame
                        * frame_samples,
                    )

                    if end_sample <= start_sample:
                        continue

                    raw_candidate_rows.append({
                        "segmentation_code":
                            segmentation_code,
                        "block_index":
                            block_index,
                        "C":
                            float(
                                C
                            ),
                        "threshold_db":
                            float(
                                threshold_db
                            ),
                        "raw_start_sample":
                            int(
                                start_sample
                            ),
                        "raw_end_sample":
                            int(
                                end_sample
                            ),
                    })

    raw_candidates = pd.DataFrame(
        raw_candidate_rows
    )

    if raw_candidates.empty:
        return (
            raw_candidates,
            pd.DataFrame(),
        )

    padded_rows = []

    for segmentation_code, group in raw_candidates.groupby(
        "segmentation_code",
        sort=True,
    ):
        pad_samples = int(
            round(
                sr
                * PADDING_MS_BY_CODE[
                    segmentation_code
                ]
                / 1000.0
            )
        )

        padded_intervals = [
            (
                max(
                    0,
                    int(
                        row.raw_start_sample
                    )
                    - pad_samples,
                ),
                min(
                    total_samples,
                    int(
                        row.raw_end_sample
                    )
                    + pad_samples,
                ),
            )
            for row in group.itertuples(
                index=False
            )
        ]

        merged = merge_intervals(
            padded_intervals
        )

        for merged_index, (
            start_sample,
            end_sample,
        ) in enumerate(
            merged
        ):
            padded_rows.append({
                "segmentation_code":
                    segmentation_code,
                "candidate_index":
                    int(
                        merged_index
                    ),
                "padding_ms_per_side":
                    int(
                        PADDING_MS_BY_CODE[
                            segmentation_code
                        ]
                    ),
                "start_sample":
                    int(
                        start_sample
                    ),
                "end_sample":
                    int(
                        end_sample
                    ),
                "start_seconds":
                    start_sample
                    / sr,
                "end_seconds":
                    end_sample
                    / sr,
                "duration_seconds":
                    (
                        end_sample
                        - start_sample
                    )
                    / sr,
            })

    candidates = pd.DataFrame(
        padded_rows
    )

    return (
        raw_candidates,
        candidates,
    )
```

## 10. Score all candidates from one field file

```python
def score_field_candidates(
    path,
    candidates,
):
    path = Path(
        path
    )

    info = sf.info(
        path
    )

    sr = int(
        info.samplerate
    )

    rows = []

    with sf.SoundFile(
        path,
        "r",
    ) as audio_file:

        for row_index, row in enumerate(
            candidates.itertuples(
                index=False
            ),
            start=1,
        ):
            start_sample = int(
                row.start_sample
            )

            end_sample = int(
                row.end_sample
            )

            audio_file.seek(
                start_sample
            )

            waveform = audio_file.read(
                frames=(
                    end_sample
                    - start_sample
                ),
                dtype="float64",
                always_2d=False,
            )

            waveform = to_mono(
                waveform
            )

            X = plp_rasta_20(
                waveform,
                sr,
            )

            decision = score_candidate_features(
                X
            )

            out = {
                "segmentation_code":
                    row.segmentation_code,
                "candidate_index":
                    int(
                        row.candidate_index
                    ),
                "padding_ms_per_side":
                    int(
                        row.padding_ms_per_side
                    ),
                "start_sample":
                    start_sample,
                "end_sample":
                    end_sample,
                "start_seconds":
                    float(
                        row.start_seconds
                    ),
                "end_seconds":
                    float(
                        row.end_seconds
                    ),
                "duration_seconds":
                    float(
                        row.duration_seconds
                    ),
                "n_plp_frames":
                    int(
                        len(X)
                    ),
                "feature_valid":
                    decision
                    is not None,
            }

            if decision is None:
                out.update({
                    "hypothesis":
                        None,
                    "hypothesis_score":
                        np.nan,
                    "llr":
                        np.nan,
                    "threshold":
                        np.nan,
                    "accepted":
                        False,
                    "prediction":
                        None,
                })

                for code_name in CLASSES:
                    out[
                        f"score_{code_name}"
                    ] = np.nan

            else:
                out.update({
                    "hypothesis":
                        decision[
                            "hypothesis"
                        ],
                    "hypothesis_score":
                        decision[
                            "hypothesis_score"
                        ],
                    "llr":
                        decision[
                            "llr"
                        ],
                    "threshold":
                        decision[
                            "threshold"
                        ],
                    "accepted":
                        decision[
                            "accepted"
                        ],
                    "prediction":
                        decision[
                            "prediction"
                        ],
                })

                for code_name in CLASSES:
                    out[
                        f"score_{code_name}"
                    ] = (
                        decision[
                            "scores"
                        ][
                            code_name
                        ]
                    )

            rows.append(
                out
            )

    return pd.DataFrame(
        rows
    )
```

## 11. Presence aggregation

For every modelled species:

- `1` = at least one accepted detection in the sample;
- `0` = no accepted detection.

For `f09`:

- `NA` = no deployable model.

That distinction matters. `NA` is epistemic limitation, not ecological absence.

```python
PAPER_CLASS_ORDER = (
    "f01",
    "f02",
    "f03",
    "f04",
    "f05",
    "f06",
    "f07",
    "f08",
    "f09",
    "f10",
)


def presence_from_detections(
    sample_id,
    detections,
):
    row = {
        "sample_id":
            sample_id,
    }

    accepted_predictions = set()

    if (
        not detections.empty
        and "accepted"
        in detections.columns
    ):
        accepted_predictions = set(
            detections.loc[
                detections[
                    "accepted"
                ].astype(bool),
                "prediction",
            ]
            .dropna()
            .astype(str)
        )

    for code_name in PAPER_CLASS_ORDER:
        column = (
            f"presence_{code_name}"
        )

        if code_name not in CLASSES:
            row[
                column
            ] = pd.NA
        else:
            row[
                column
            ] = int(
                code_name
                in accepted_predictions
            )

    return row
```

## 12. Deduplicate accepted detections for audit summaries

The same acoustic event can be proposed by several overlapping segmentation bands. Raw decisions are
preserved, but a second table merges overlapping **accepted detections of the same predicted species**.

Presence/absence itself is unaffected either way because one accepted call is sufficient.

```python
def deduplicate_accepted_detections(
    detections,
):
    if detections.empty:
        return pd.DataFrame()

    accepted = (
        detections[
            detections[
                "accepted"
            ].astype(bool)
            & detections[
                "prediction"
            ].notna()
        ]
        .copy()
    )

    if accepted.empty:
        return pd.DataFrame()

    rows = []

    for prediction, group in accepted.groupby(
        "prediction",
        sort=True,
    ):
        group = group.sort_values(
            [
                "start_sample",
                "end_sample",
            ]
        )

        current = None

        for row in group.itertuples(
            index=False
        ):
            if current is None:
                current = {
                    "prediction":
                        prediction,
                    "start_sample":
                        int(
                            row.start_sample
                        ),
                    "end_sample":
                        int(
                            row.end_sample
                        ),
                    "best_llr":
                        float(
                            row.llr
                        ),
                    "threshold":
                        float(
                            row.threshold
                        ),
                    "segmentation_codes":
                        {
                            str(
                                row.segmentation_code
                            )
                        },
                    "raw_detection_count":
                        1,
                }

                continue

            if int(
                row.start_sample
            ) <= current[
                "end_sample"
            ]:
                current[
                    "end_sample"
                ] = max(
                    current[
                        "end_sample"
                    ],
                    int(
                        row.end_sample
                    ),
                )

                current[
                    "best_llr"
                ] = max(
                    current[
                        "best_llr"
                    ],
                    float(
                        row.llr
                    ),
                )

                current[
                    "segmentation_codes"
                ].add(
                    str(
                        row.segmentation_code
                    )
                )

                current[
                    "raw_detection_count"
                ] += 1

            else:
                rows.append(
                    current
                )

                current = {
                    "prediction":
                        prediction,
                    "start_sample":
                        int(
                            row.start_sample
                        ),
                    "end_sample":
                        int(
                            row.end_sample
                        ),
                    "best_llr":
                        float(
                            row.llr
                        ),
                    "threshold":
                        float(
                            row.threshold
                        ),
                    "segmentation_codes":
                        {
                            str(
                                row.segmentation_code
                            )
                        },
                    "raw_detection_count":
                        1,
                }

        if current is not None:
            rows.append(
                current
            )

    out = pd.DataFrame(
        rows
    )

    if out.empty:
        return out

    out[
        "segmentation_codes"
    ] = out[
        "segmentation_codes"
    ].apply(
        lambda values:
            ",".join(
                sorted(
                    values
                )
            )
    )

    return out
```

## 13. Per-file checkpoint paths

```python
def safe_file_stem(
    sample_id,
):
    return re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        sample_id,
    )


def checkpoint_paths(
    sample_id,
):
    stem = safe_file_stem(
        sample_id
    )

    return {
        "raw_candidates":
            PER_FILE_DIR
            / f"{stem}__raw_candidates.csv",
        "candidates":
            PER_FILE_DIR
            / f"{stem}__candidates.csv",
        "detections":
            PER_FILE_DIR
            / f"{stem}__detections.csv",
        "accepted_deduplicated":
            PER_FILE_DIR
            / f"{stem}__accepted_deduplicated.csv",
        "presence":
            PER_FILE_DIR
            / f"{stem}__presence.csv",
        "run":
            PER_FILE_DIR
            / f"{stem}__run.json",
    }


def checkpoint_complete(
    sample_id,
):
    paths = checkpoint_paths(
        sample_id
    )

    if not paths[
        "run"
    ].exists():
        return False

    try:
        with open(
            paths[
                "run"
            ],
            "r",
            encoding="utf-8",
        ) as f:
            run = json.load(
                f
            )

        return (
            run.get(
                "status"
            )
            == "ok"
            and paths[
                "presence"
            ].exists()
            and paths[
                "detections"
            ].exists()
        )

    except Exception:
        return False
```

```python
# ============================================================
# GPU opcional para la segmentación del notebook 09
# PONER ESTA CELDA JUSTO ANTES DE: "14. Run field inference"
#
# Si encuentra CUDA + CuPy:
#   FIR -> GPU
#   STE -> GPU
#   Moving Average -> GPU
#   log-energy -> GPU
#
# Si no encuentra GPU/CuPy:
#   conserva automáticamente las funciones CPU originales.
#
# Mantiene float64 para no cambiar innecesariamente los resultados.
# ============================================================

# Guardar las implementaciones CPU originales una sola vez.
if "_CPU_bandpass_for_segmentation" not in globals():
    _CPU_bandpass_for_segmentation = bandpass_for_segmentation
    _CPU_short_time_energy = short_time_energy
    _CPU_moving_average_same = moving_average_same
    _CPU_energy_to_db = energy_to_db


USE_GPU = False
GPU_NAME = None

try:
    import cupy as cp
    from cupyx.scipy.signal import fftconvolve as gpu_fftconvolve

    if cp.cuda.runtime.getDeviceCount() > 0:
        USE_GPU = True

        device_id = cp.cuda.Device().id

        GPU_NAME = (
            cp.cuda.runtime
            .getDeviceProperties(device_id)["name"]
        )

        if isinstance(GPU_NAME, bytes):
            GPU_NAME = GPU_NAME.decode()

except Exception as exc:
    USE_GPU = False
    print("GPU/CuPy no disponible:", repr(exc))


if USE_GPU:

    print("=" * 60)
    print("GPU ACTIVADA")
    print("Dispositivo:", GPU_NAME)
    print("=" * 60)

    # --------------------------------------------------------
    # Cachés GPU
    # --------------------------------------------------------

    _GPU_FIR_CACHE = {}

    # Mantener el bloque actual en GPU para que las 9 bandas
    # no vuelvan a copiar los mismos ~30 s desde CPU.
    _GPU_LAST_CPU_BLOCK = None
    _GPU_LAST_BLOCK = None


    def _get_gpu_audio_block(audio):
        global _GPU_LAST_CPU_BLOCK
        global _GPU_LAST_BLOCK

        # Mismo objeto NumPy durante las 9 especies del bloque.
        if audio is not _GPU_LAST_CPU_BLOCK:

            _GPU_LAST_CPU_BLOCK = audio

            _GPU_LAST_BLOCK = cp.asarray(
                np.asarray(
                    audio,
                    dtype=np.float64,
                )
            )

        return _GPU_LAST_BLOCK


    # --------------------------------------------------------
    # Reemplazo GPU: FIR
    # --------------------------------------------------------

    def bandpass_for_segmentation(
        audio,
        sr,
        low_hz,
        high_hz,
    ):
        key = (
            int(sr),
            float(low_hz),
            float(high_hz),
            int(FIR_NUMTAPS),
        )

        # FIR se diseña una sola vez.
        if key not in _GPU_FIR_CACHE:

            # Reutilizar el FIR CPU si ya existe.
            if key not in _FIR_CACHE:
                _FIR_CACHE[key] = firwin(
                    numtaps=FIR_NUMTAPS,
                    cutoff=[
                        low_hz,
                        high_hz,
                    ],
                    pass_zero=False,
                    fs=sr,
                )

            _GPU_FIR_CACHE[key] = cp.asarray(
                _FIR_CACHE[key],
                dtype=cp.float64,
            )

        gpu_audio = _get_gpu_audio_block(
            audio
        )

        return gpu_fftconvolve(
            gpu_audio,
            _GPU_FIR_CACHE[key],
            mode="same",
        )


    # --------------------------------------------------------
    # Reemplazo GPU: STE
    # --------------------------------------------------------

    def short_time_energy(
        audio,
        sr,
    ):
        # Si por alguna razón llega NumPy, usar CPU.
        if not isinstance(
            audio,
            cp.ndarray,
        ):
            return _CPU_short_time_energy(
                audio,
                sr,
            )

        frame_samples = int(
            round(
                sr
                * STE_FRAME_SECONDS
            )
        )

        n_frames = (
            int(audio.size)
            // frame_samples
        )

        if n_frames == 0:
            return (
                cp.empty(
                    0,
                    dtype=cp.float64,
                ),
                frame_samples,
            )

        trimmed = audio[
            :
            n_frames
            * frame_samples
        ]

        frames = trimmed.reshape(
            n_frames,
            frame_samples,
        )

        energy = cp.sum(
            frames
            * frames,
            axis=1,
        )

        return (
            energy,
            frame_samples,
        )


    # --------------------------------------------------------
    # Reemplazo GPU: moving average MA=12
    # --------------------------------------------------------

    def moving_average_same(
        values,
    ):
        if not isinstance(
            values,
            cp.ndarray,
        ):
            return _CPU_moving_average_same(
                values
            )

        if values.size == 0:
            return values.copy()

        kernel = (
            cp.ones(
                STE_MA_LENGTH,
                dtype=cp.float64,
            )
            / STE_MA_LENGTH
        )

        return cp.convolve(
            values,
            kernel,
            mode="same",
        )


    # --------------------------------------------------------
    # Reemplazo GPU: conversión a dB
    #
    # AQUÍ regresamos a NumPy porque ste_db tiene solo ~3000
    # valores por bloque. El pesado FIR/STE ya ocurrió en GPU.
    # Así endpoint_threshold_db y find_endpoint_runs quedan
    # exactamente como estaban.
    # --------------------------------------------------------

    def energy_to_db(
        energy,
    ):
        if not isinstance(
            energy,
            cp.ndarray,
        ):
            return _CPU_energy_to_db(
                energy
            )

        tiny = np.finfo(
            np.float64
        ).tiny

        energy_db_gpu = (
            10.0
            * cp.log10(
                cp.maximum(
                    energy,
                    tiny,
                )
            )
        )

        return cp.asnumpy(
            energy_db_gpu
        )


    # Limpiar memoria GPU retenida de ejecuciones anteriores.
    cp.get_default_memory_pool().free_all_blocks()
    cp.get_default_pinned_memory_pool().free_all_blocks()

    print(
        "Funciones de segmentación sustituidas "
        "por versiones GPU."
    )


else:

    # Restaurar CPU si se vuelve a ejecutar la celda
    # en una máquina sin CUDA.
    bandpass_for_segmentation = (
        _CPU_bandpass_for_segmentation
    )

    short_time_energy = (
        _CPU_short_time_energy
    )

    moving_average_same = (
        _CPU_moving_average_same
    )

    energy_to_db = (
        _CPU_energy_to_db
    )

    print("=" * 60)
    print("GPU NO DISPONIBLE")
    print("Se continuará con CPU.")
    print("=" * 60)
```

## 15. Assemble the field presence table from checkpoints

```python
presence_blocks = []

if RUN_FIELD_SCAN:

    successful_ids = (
        field_run_log.loc[
            field_run_log[
                "status"
            ] == "ok",
            "sample_id",
        ]
        .astype(str)
        .tolist()
    )

    for sample_id in successful_ids:
        path = checkpoint_paths(
            sample_id
        )[
            "presence"
        ]

        presence_blocks.append(
            pd.read_csv(
                path
            )
        )

field_presence = (
    pd.concat(
        presence_blocks,
        ignore_index=True,
    )
    if presence_blocks
    else pd.DataFrame()
)

if not field_presence.empty:
    metadata_cols = (
        eligible_inventory[
            [
                "sample_id",
                "relative_path",
                "pond",
                "month",
                "duration_seconds",
            ]
        ]
    )

    field_presence = (
        metadata_cols.merge(
            field_presence,
            on="sample_id",
            how="inner",
            validate="one_to_one",
        )
    )

    for code_name in CLASSES:
        col = (
            f"presence_{code_name}"
        )

        field_presence[
            col
        ] = (
            field_presence[
                col
            ]
            .astype(
                "Int64"
            )
        )

    field_presence[
        "presence_f09"
    ] = pd.array(
        [
            pd.NA
        ]
        * len(
            field_presence
        ),
        dtype="Int64",
    )

    field_presence.to_csv(
        RESULTS_DIR
        / "field_presence_absence.csv",
        index=False,
    )

    display(
        field_presence.head(
            30
        )
    )

    print(
        "Presence rows:",
        len(
            field_presence
        ),
    )
```

## 16. Build global event tables

Raw per-file checkpoints remain the primary audit trail.

This optional consolidation is convenient for analysis.

```python
field_candidates = pd.DataFrame()
field_detections = pd.DataFrame()
field_accepted_deduplicated = pd.DataFrame()

if (
    RUN_FIELD_SCAN
    and BUILD_GLOBAL_EVENT_TABLES
):

    candidate_blocks = []
    detection_blocks = []
    accepted_blocks = []

    for sample_id in (
        field_run_log.loc[
            field_run_log[
                "status"
            ] == "ok",
            "sample_id",
        ]
        .astype(str)
    ):
        paths = checkpoint_paths(
            sample_id
        )

        for (
            path_key,
            destination,
        ) in (
            (
                "candidates",
                candidate_blocks,
            ),
            (
                "detections",
                detection_blocks,
            ),
            (
                "accepted_deduplicated",
                accepted_blocks,
            ),
        ):
            path = paths[
                path_key
            ]

            if (
                path.exists()
                and path.stat().st_size
                > 0
            ):
                try:
                    frame = pd.read_csv(
                        path
                    )

                    if not frame.empty:
                        destination.append(
                            frame
                        )
                except pd.errors.EmptyDataError:
                    pass

    field_candidates = (
        pd.concat(
            candidate_blocks,
            ignore_index=True,
        )
        if candidate_blocks
        else pd.DataFrame()
    )

    field_detections = (
        pd.concat(
            detection_blocks,
            ignore_index=True,
        )
        if detection_blocks
        else pd.DataFrame()
    )

    field_accepted_deduplicated = (
        pd.concat(
            accepted_blocks,
            ignore_index=True,
        )
        if accepted_blocks
        else pd.DataFrame()
    )

    field_candidates.to_csv(
        RESULTS_DIR
        / "field_candidates.csv",
        index=False,
    )

    field_detections.to_csv(
        RESULTS_DIR
        / "field_detector_decisions.csv",
        index=False,
    )

    field_accepted_deduplicated.to_csv(
        RESULTS_DIR
        / "field_accepted_detections_deduplicated.csv",
        index=False,
    )

    print("Global candidates :", len(field_candidates))
    print("Global decisions  :", len(field_detections))
    print(
        "Deduplicated accepted:",
        len(
            field_accepted_deduplicated
        ),
    )
```

## 19. Detector-output sanity checks

These checks look for pathological field behavior. They do not constitute test accuracy.

```python
if not field_presence.empty:

    for code_name in CLASSES:
        column = (
            f"presence_{code_name}"
        )

        unique = set(
            field_presence[
                column
            ]
            .dropna()
            .astype(int)
            .unique()
        )

        if not unique.issubset(
            {
                0,
                1,
            }
        ):
            raise AssertionError(
                f"Non-binary presence values for {code_name}: {unique}"
            )

    if not field_presence[
        "sample_id"
    ].is_unique:
        raise AssertionError(
            "Presence sample_id must be unique."
        )

    if not field_presence[
        "presence_f09"
    ].isna().all():
        raise AssertionError(
            "Unmodelled f09 must remain NA."
        )

    print("Binary presence for modelled classes : OK")
    print("Unique sample IDs                    : OK")
    print("f09 is NA, not zero                 : OK")

if (
    not field_detections.empty
):
    valid = field_detections[
        field_detections[
            "feature_valid"
        ].astype(bool)
    ]

    print()
    print(
        "Valid feature fraction:",
        round(
            field_detections[
                "feature_valid"
            ].astype(bool).mean(),
            4,
        ),
    )

    if not valid.empty:
        print(
            "Raw accepted fraction:",
            round(
                valid[
                    "accepted"
                ].astype(bool).mean(),
                4,
            ),
        )

        print(
            "Median candidate duration:",
            round(
                valid[
                    "duration_seconds"
                ].median(),
                4,
            ),
            "s",
        )
```

## 24. Final interpretation

After a successful full run, inspect in this order:

1. `field_run_log`
2. `field_presence`
3. `presence_summary`
4. `accepted_cross_tab`
5. `llr_field_summary`

### What counts as success

A successful execution means:

- the recovered Dataset-2 inventory has been explicitly audited;
- all intended eligible files are processed without hidden failures;
- frozen segmentation and frozen detector artifacts are used unchanged;
- accepted calls are aggregated to presence/absence;
- `f09` remains `NA`;
- results are saved with per-file audit trails.

### What does **not** count as success

A large number of detections is not automatically evidence that the system works well. Pond audio
contains unmodelled frogs, insects, human sounds, mechanical artifacts and environmental noise. The
verification layer is supposed to reject much of that, but independent labels are required to quantify
how well it actually does so.

### Conceptual status of the paper replication

With this notebook, the full conceptual pipeline of the source article has been implemented:

**labelled calls → acoustic features → GMM species models → likelihood-ratio verification → automatic
call segmentation → field presence/absence.**

Anything after this point is evaluation, ecological summarization, error analysis, or methodological
extension rather than another missing major block of the original pipeline.
